# Solution

Start with the main solution

```python
def count_duplicates(nums):

    counts = dict()
    for x in nums:
        if x in counts:
            counts[x] = counts[x]+1
        else:
            counts[x] = 1

    counter = 0
    for key, value in counts.items():
        if value > 1:
            counter += 1

    return counter
```
